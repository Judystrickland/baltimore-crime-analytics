```python
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv(r"C:\Users\chyj2\OneDrive\Desktop\baltimore-crime-analytics\data\cleaned\cleaned_crime_data.csv")
```

```python
print(df.columns)
```

```python
#Crime Type Analysis This step quickly counts how many times each specific 
#type of crime (e.g., Robbery, Larceny) appears in the entire dataset.
    # Count incidents by crime description
crime_type = df["description"].value_counts().reset_index()

# Rename the columns for clarity
crime_type.columns = ["crime_type", "total"]

crime_type.head()
```

```python
## Monthly crime Agregation
# Group data by year and month to get total monthly volume
monthly_crime = df.groupby(["year", "month"]).size().reset_index(name="total_crimes")

# View the first few rows to verify
monthly_crime.head()
```

```python
# 1. Print the column names to see the exact spelling
print("Available columns:", df.columns.tolist())

# Drop rows with missing geographic coordinates
hotspots = df.dropna(subset=["latitude", "longitude"])

# 2. Update this list with the EXACT column names printed above
columns_to_keep = ["description", "new_district", "neighborhood", "latitude", "longitude", "crimedatetime"]

# Filter down to only the columns needed for the map
hotspots = hotspots[columns_to_keep]

# Save the cleaned data to a CSV file
hotsports.to_csv("outputs/tables/crime_hotspots.csv", index=False)
```
